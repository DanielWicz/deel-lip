# Demo 3: Exporting to Vanilla PyTorch Modules


Lipschitz models can be exported to standard `torch.nn` modules for deployment. The exported model preserves predictions while removing the training-time constraints.


```python
import torch
from deel.lip.layers import SpectralDense, GroupSort2
from deel.lip.model import Sequential, vanillaModel
```

```python
torch.manual_seed(11)
model = Sequential(
    SpectralDense(8, 16, activation="relu", use_bias=False),
    GroupSort2(),
    SpectralDense(16, 4, use_bias=False),
)
inputs = torch.randn(2, 8)
logits = model(inputs)
vanilla = vanillaModel(model)
with torch.no_grad():
    diff = (logits - vanilla(inputs)).abs().max().item()
print(f"Max prediction difference after export: {diff:.2e}")
```

Use `vanillaModel` once training is complete to benefit from faster inference while keeping the certified weights generated during optimisation.
