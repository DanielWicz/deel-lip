# Demo 0: Inspecting Spectral Normalization


Spectral layers expose kernels whose singular values are constrained. We can verify that property using `compute_layer_sv`. The resulting bounds are useful to monitor drift during training.


```python
import torch
from deel.lip.layers import SpectralDense
from deel.lip.compute_layer_sv import compute_layer_sv
```

```python
torch.manual_seed(7)
layer = SpectralDense(16, use_bias=False)
inputs = torch.randn(10, 16)
_ = layer(inputs)  # build the layer
layer.condense()
sigma_min, sigma_max = compute_layer_sv(layer.vanilla_export())
print(f"Singular values in [ {sigma_min:.4f}, {sigma_max:.4f} ]")
```

Calling `condense` freezes the trained parameters into their constrained form. Exporting to `torch.nn.Linear` with `vanilla_export` keeps compatibility with downstream tooling.
