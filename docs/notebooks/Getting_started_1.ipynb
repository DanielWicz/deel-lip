# Getting Started 1: Building a 1-Lipschitz MLP with PyTorch


This notebook shows how to assemble a simple fully-connected network using `deel-lip`'s spectral layers. The resulting model stays 1-Lipschitz by construction and interoperates seamlessly with standard `torch.nn` modules.


```python
import torch
from deel.lip.layers import SpectralDense, GroupSort2
from deel.lip.model import Sequential
```

```python
torch.manual_seed(0)

model = Sequential(
    SpectralDense(784, 128, activation="relu", use_bias=False),
    GroupSort2(),
    SpectralDense(128, 10, use_bias=False),
)
print(model)
```

```python
x = torch.randn(4, 784)
logits = model(x)
print(f"Output shape: {logits.shape}")
```

Each spectral layer keeps its weight matrix 1-Lipschitz. Composing them inside `Sequential` preserves this property, enabling provable robustness checks later in the pipeline.
